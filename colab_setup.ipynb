## 3. Explorar a estrutura do projeto

```python
import os

for root, dirs, files in os.walk('.'):
    dirs[:] = [d for d in dirs if d != '.git']
    level = root.replace('.', '').count(os.sep)
    indent = '  ' * level
    print(f'{indent}{os.path.basename(root)}/')
    for f in files:
        print(f'{indent}  {f}')
```
